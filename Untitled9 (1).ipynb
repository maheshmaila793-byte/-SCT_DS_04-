```python
# ============================
# TASK 4: Traffic Accident Analysis
# ============================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload and Load Dataset
uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("✅ Dataset loaded successfully")
print(df.head())
print(df.info())

# Step 2: Preprocessing
# Combine date and time into a single datetime column
df['datetime'] = pd.to_datetime(df['date'] + " " + df['time'])
df['hour'] = df['datetime'].dt.hour

# Categorize time of day
df['time_of_day'] = pd.cut(
    df['hour'],
    bins=[0,6,12,18,24],
    labels=['Night','Morning','Afternoon','Evening'],
    right=False
)

# Step 3: Pattern Analysis
print("\nAccidents by Road Condition:\n", df['road_condition'].value_counts())
print("\nAccidents by Weather:\n", df['weather'].value_counts())
print("\nAccidents by Time of Day:\n", df['time_of_day'].value_counts())

# Step 4: Visualizations

# Road condition vs accidents
plt.figure(figsize=(6,4))
sns.countplot(x="road_condition", data=df, hue="road_condition", palette="Set2", legend=False)
plt.title("Accidents by Road Condition")
plt.show()

# Weather vs accidents
plt.figure(figsize=(6,4))
sns.countplot(x="weather", data=df, hue="weather", palette="Set3", legend=False)
plt.title("Accidents by Weather")
plt.show()

# Time of day vs accidents
plt.figure(figsize=(6,4))
sns.countplot(x="time_of_day", data=df, hue="time_of_day", palette="Set1", legend=False)
plt.title("Accidents by Time of Day")
plt.show()

# Accident hotspots (locations)
plt.figure(figsize=(8,4))
sns.countplot(x="location", data=df, hue="location", order=df['location'].value_counts().index, palette="pastel", legend=False)
plt.title("Accident Hotspots by Location")
plt.xticks(rotation=45)
plt.show()

# Step 5: Contributing Factors

# Severity distribution
plt.figure(figsize=(6,4))
sns.countplot(x="severity", data=df, hue="severity", palette="coolwarm", legend=False)
plt.title("Accidents by Severity")
plt.show()

# Vehicle type vs severity
plt.figure(figsize=(6,4))
sns.countplot(x="vehicle_type", hue="severity", data=df, palette="muted")
plt.title("Severity by Vehicle Type")
plt.show()

# Alcohol involvement
plt.figure(figsize=(6,4))
sns.countplot(x="alcohol_involved", data=df, hue="alcohol_involved", palette="RdBu", legend=False)
plt.title("Accidents with Alcohol Involvement")
plt.show()

# Police reports
plt.figure(figsize=(6,4))
sns.countplot(x="police_report", data=df, hue="police_report", palette="Greens", legend=False)
plt.title("Accidents Reported to Police")
plt.show()
```
